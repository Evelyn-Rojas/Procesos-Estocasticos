**¿Qué es una caminata aleatoria?**

Es un proceso estocástico que modela una trayectoria consistente en una sucesión de pasos aleatorios, donde la posición futura depende únicamente de la actual más un cambio impredecible

```python
import random as r
```

```python
pb=r.random()
print(pb)
```

```python
def caminata(pb):
  u=r.random()
  if u<pb:
    return 1
  else:
    return -1
```

```python
pasos=[]
sum=0
n=100
l=[0]
for i in range(n):
  pasos.append(caminata(pb))
  sum+=caminata(pb)
  l.append(sum)
#print(pasos)
#print(sum)
```

```python
import matplotlib.pyplot as plt
plt.plot(l, color="pink")
plt.xlabel("pasos")
plt.ylabel("distancia")
plt.show()
```

```python
def caminata2(pb):
  pasos=[]
  sum=0
  n=10**3
  l=[0]
  for i in range(n):
    pasos.append(caminata(pb))
    sum+=caminata(pb)
    l.append(sum)
  plt.plot(l, color="pink")
  plt.xlabel("pasos")
  plt.ylabel("distancia")
  plt.show()
caminata2(pb)
```

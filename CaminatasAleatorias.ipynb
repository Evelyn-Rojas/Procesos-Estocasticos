**¿Qué es una caminata aleatoria?**

Es un proceso estocástico que modela una trayectoria consistente en una sucesión de pasos aleatorios, donde la posición futura depende únicamente de la actual más un cambio impredecible

$$
\text{Parte de}
$$

$$
X_0 = 0
$$

En cada paso:

$$
P(\text{suma } +1) = p
$$

$$
P(\text{suma } -1) = 1 - p
$$

Modelo matemático:

$$
X_n = X_{n-1} + \varepsilon_n
$$

donde:

$$
\varepsilon_n \text{ es una variable aleatoria independiente}
$$

Caso simétrico:

$$
P(\varepsilon_n = 1) = 0.5
$$

$$
P(\varepsilon_n = -1) = 0.5
$$


```python
import random as r #Importamos la libreria random para generar números aleatorios
```

```python
pb=r.random() #Probabilidad de p, usamos random para irla cambiando y sea aleatoria
print(pb)
```

```python
#Utilizamos la siguiente función para definir la caminata
def caminata(pb):
  u=r.random()
  if u<pb:
    return 1
  else:
    return -1
```

```python
pasos=[]
sum=0 #Variable para la distancia final
n=100 #númera de iteraciones
l=[0]
for i in range(n):
  pasos.append(caminata(pb))
  sum+=caminata(pb)
  l.append(sum)
#print(pasos)
#print(sum)
```

```python
import matplotlib.pyplot as plt #Libreria para gráficar
plt.plot(l, color="pink")
plt.xlabel("Pasos")
plt.ylabel("Distancia")
plt.show()
```

```python
def caminata2(pb):
  pasos=[]
  sum=0
  n=10**3
  l=[0]
  for i in range(n):
    pasos.append(caminata(pb))
    sum+=caminata(pb)
    l.append(sum)
  plt.plot(l, color="pink")
  plt.xlabel("pasos")
  plt.ylabel("distancia")
  plt.show()
caminata2(pb)
```

```python
from statistics import variance #Importamos la libreria para obtener la varianza de las distancias
def caminata2(pb):
  pasos=[]
  sum=0
  n=10**2
  l=[0]
  l22=[]
  for i in range(n):
    pasos.append(caminata(pb))
    sum+=caminata(pb)
    l.append(sum)
    l22.append(abs(sum))
  valor=variance(l22)
  print("Varianza:", valor)
caminata2(pb)
```

Todos nuestros resultados obtenidos están sujetos al número de veces que repetimos el proceso, por lo cual al aumentar o disminuir $n$  habrá cambios, tomando en cuenta que probabilidad tendrá tanto p como q.

En conclusión, aunque cada realización es aleatoria, el comportamiento global del sistema está completamente determinado por
$p$ y $n$.
